# **1.GET Request**

````python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

print("Status Code:", response.status_code)
print(response.text)
````

# **2.Understanding JSON**

````python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/todos/1")

data = response.json()

print(data)
print(type(data))
print(data["title"])
````

# **3.POST Request**

URL:
https://postman-echo.com/post


````python
import requests

payload = {"name": "smith",
           "company": "xyz",
           "role": "AI Engineer"}

response = requests.post("https://postman-echo.com/post", json=payload, timeout=30)

print("Status Code:", response.status_code)
print(response.json())
````

# **5. Understanding the Engineering Problem**

````python
from openai import OpenAI

client = OpenAI(api_key="your_api_key")

question = input("Ask something: ")

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role":"user",
            "content":question
        }
    ]
)

print(response.choices[0].message.content)
````
